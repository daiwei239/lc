```python
def minCost(self, startPos: List[int], homePos: List[int], rowCosts: List[int], colCosts: List[int]) -> int:
    m = len(rowCosts)
    n = len(colCosts)
    dp = [[inf]* n for _ in range(m)]
    dp[0][0] = 0
    for i in range(m):
        for j in range(n):
            for ii,jj in [[-1,0],[1,0],[0,-1],[0,1]]:
                x,y = i-ii,j-jj
                if 0<=x<m and 0<=y<n:
                    if jj ==0:
                        dp[x][y] = min(dp[x][y],dp[i][j]+rowCosts[x])
                    elif ii ==0:
                        dp[x][y] = min(dp[x][y],dp[i][j]+colCosts[y])
                
```
