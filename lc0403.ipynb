```python
from typing import List
```

```python
from bisect import bisect_left,bisect_right


def maxWalls(self, robots: List[int], distance: List[int], walls: List[int]) -> int:
    walls.sort()
    a = [(0,0)]+sorted(zip(robots,distance),key= lambda x:x[0])+[(inf,0)]
    n = len(robots)
    f = [(0,0) for _ in range(n+1)]
    for i in range(1,n+1):
        x1,d = a[i]
        # 当前i机器人向左射击
        left_x = max(x1-d,a[i-1][0]+1)
        left = bisect_left(walls,left_x)
        curr = bisect_right(walls,x1)
        # 这个地方为什么要用到f[i-1][0],因为这个就是假设i左射击的结果，而当前i就是要左射击
        left_res = f[i-1][0]+curr-left
        # 当前i机器人向右射击
        # 对于和机器人相同位置的墙，左击穿和右击穿都可以，但是只会选一个，所以都需要有
        curr = bisect_left(walls,x1)
        for j in range(2):
            x2,d2 = a[i+1]
            if j == 0:
                x2-=d2
            right_x = min(x2-1,x1+d)
            right = bisect_right(walls,right_x)
            f[i][j] = max(left_res,f[i-1][1]+right-curr)
    return f[n][1]
```

# 题目**239**✨✨✨

```python
import heapq
```

```python
def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
    n = len(nums)
    if n<k:
        return []
    heap =[]
    for i in range(k-1):
        heap.append((-nums[i],i))
    heapq.heapify(heap)
    
    res = []

    for i in range(k-1,n):
        heapq.heappush(heap,(-nums[i],i))
        while heap[0][1]<i-k+1:
            heapq.heappop(heap)

        res.append(heap[0][0])
    return res

    
```

## 方法二，采用双端队列的方法

```python
from collections import deque
```

```python
def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
    queue = deque()
    n = len(nums)
    res = []
    for i in range(k):
        while queue and nums[i]>=nums[queue[-1]]:
            queue.pop()
        queue.append(i)
    res.append(nums[queue[0]])

    for j in range(k,n):
        while queue and nums[j]>=nums[queue[-1]]:
            queue.pop()
        queue.append(j)
        while queue[0]<=j-k:
            queue.popleft()
        res.append(nums[queue[0]])
    return res
```

# 题目**84**✨✨✨

```python
def largestRectangleArea(self, heights: List[int]) -> int:
    heights = [0]+heights+[0]
    ans = 0
    stack = []
    for right,val in enumerate(heights):
        while stack and stack[-1]>val:
            i = stack.pop()
            ans  = max(ans,(right-stack[-1]-1)*heights[i])
        stack.append(i)
    return ans
```

# 题目**295**✨✨✨

```python
class MedianFinder:

    def __init__(self):
        self.min_heap =[]
        self.max_heap = []
        heapq.heapify(self.min_heap)
        heapq.heapify(self.max_heap)
             
    def addNum(self, num: int) -> None:
        if len(self.min_heap) == len(self.max_heap):
            if num>self.min_heap[0]:
                heapq.heappush(self.min_heap,num)
                temp = heapq.heappop(self.min_heap)
                heapq.heappush(self.max_heap,-temp)
            else:
                heapq.heappush(self.max_heap,-num)
        else:
            if num>=-self.max_heap[0]:
                heapq.heappush(self.min_heap,num)
            else:
                heapq.heappush(self.max_heap,-num)
                temp = -heapq.heappop(self.max_heap)
                heapq.heappush(self.min_heap,temp)

    def findMedian(self) -> float:
        if len(self.min_heap) == len(self.max_heap):
            return (heapq.heappop(self.min_heap)-heapq.heappop(self.max_heap))//2
        else:
            return -heapq.heappop(self.max_heap)
```

# 题目**25**✨✨✨

```python
from typing import Optional
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
    def reverse(head:Optional[ListNode],k:int)->Optional[ListNode]:
        prev = None
        curr = head
        i = 0
        while curr and i<k:
            nxt = curr.next
            curr.next = prev
            prev = curr
            curr = nxt
            i+=1
        head.next = curr

        return prev
    dummy = ListNode(-1)
    dummy.next = head
    temp = head
    total = 0
    while temp:
        total+=1
        temp = temp.next
    total//=k
    temp = dummy
    for _ in range(total):
        old = temp.next
        newHead=reverse(old,k)
        temp.next = newHead
        temp = old
    return dummy.next
```

# 题目**206**✨

```python
def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
    prev = None
    curr = head
    while curr:
        nxt = curr.next
        curr.next = prev
        prev = curr
        curr = nxt
    return prev
```
